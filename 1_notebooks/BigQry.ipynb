# 1.- **Autenticación y conexión a BigQuery**

```python
#Conéctate a BigQuery desde Colab
from google.colab import auth
auth.authenticate_user()

#Luego activa BigQuery:
from google.cloud import bigquery
import pandas as pd

project_id = "fintech-ml-portfolio"  #ID real de proyecto
client = bigquery.Client(project=project_id)
```

```python
query = """
  SELECT
    *
  FROM
    `bigquery-public-data.world_bank_intl_education.international_education`
  WHERE indicator_code = 'NY.GDP.PCAP.CD'
    AND year BETWEEN 2010 AND 2020
  ORDER BY year DESC
  LIMIT 10
"""
df_bq = client.query(query).to_dataframe()
df_bq.head()
```
